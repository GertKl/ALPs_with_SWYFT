```python
import os
import sys
import numpy as np
import swyft
import matplotlib.pyplot as plt
import copy
sys.path.append('/home/gertwk/ALPs_with_SWYFT/analysis_scripts/ALP_sim')
```

```python
main_dir_name = "ALPs_with_SWYFT"
main_dir = os.getcwd().split(main_dir_name)[0]+main_dir_name+"/"
thesis_figs_dir = main_dir+"thesis_figures/residual_logs/"
os.makedirs(thesis_figs_dir,exist_ok=True)
irf_dir = main_dir+"IRFs/"
scripts_dir = main_dir+"analysis_scripts/ALP_sim/" # The directory where ALP_quick_sim.py is placed. Can be empty if it's in same folder as notebook.
```

```python
irf_file = irf_dir+"CTA/Prod5-North-20deg-AverageAz-4LSTs09MSTs.180000s-v0.1.fits"    # The path to the IRF_file (relative to this notebook)
```

```python
if scripts_dir : sys.path.append(scripts_dir ) 
from ALP_quick_sim import ALP_sim
```

```python
A = ALP_sim(set_null=False,set_obs=False)
```

```python
A.configure_obs(
    signal = True,
    edisp = True,
    bkg = True,
    emin = 1e1,
    emax = 1e4,
    nbins = 40, #90
    nbins_etrue = 180,
    pointing = [150.58,-13.26],
    livetime=1,
    irf_file = irf_file,
    geom = 'gamma',    
)

A.configure_model(
    model = "",
    noise = "poisson",
    params = [0,0],
    log_params = [0,0],
    null_params = [0,0],  # should have been called 'standard_params', oh well...
    signal = True,
    bkg = True,
    logcounts = False,
    residuals = False,
    floor = None,
    floor_obs = None,
    ALP_seed = 0,
    noise_seed = 0,        # Makes sure the noise always turns out the same
)

A.configure_plot(
    xmin = None,
    xmax = None,
    ymin = None,
    ymax = None,
    logx = True,
    logy = True,
    legend = True,
    dnde = False,
)
```

```python
# A simple test 
A.compute_case(
    new_fig = True,
    new_counts = True,
    null = False,
    plot_obs = True,
    plot_exp = True,
    plot_survival = True,
    errorbands = True,
    errors = True,
    linestyle = "-",
    linestyle_obs = "-", 
    label_exp = True,
    label_obs = True,
    label_survival = True,
    transparency = 0,
    axes = None,
    axes_survival = None,
)
```

```python
# Seed for B-field configurations, and [m,g] to be reused for several plots
demo_ALP_seed = 0
demo_params = [50,1]
xmin = 9e0
xmax = 1.1e4

# General arguments to compute_case(), to be re-used in most cases
general_arguments = dict(
    new_fig = False,
    new_counts = True,
    null = False,
    plot_obs = True,
    plot_exp = True,
    plot_survival = False,
    errorbands = True,
    errors = True,
    linestyle = "-",
    linestyle_obs = "-", 
    label_exp = True,
    label_obs = True,
    label_survival = True,
    transparency = 0,
    axes = None,
    axes_survival = None,
    color = 'r',
)

# Arguments to compute_case() to replace some of the general arguments when plotting the standard hypothesis
standard_hypothesis_arguments = dict(
    new_fig=True,
    null=True,
    plot_obs=False,
    color='k',
)

# General arguments to be used for plotting the photon survival probability
pgg_arguments = dict(
    new_fig = False,
    new_counts = True,
    null = False,
    plot_obs = False,
    plot_exp = False,
    plot_survival = True,
    errorbands = True,
    errors = True,
    linestyle = "-", 
    label_survival = True,
    transparency = 0,
    axes_survival = None,
    color = 'r',
)
```

```python
fig_width = 9
quadruple_fig = plt.figure(figsize=(fig_width,fig_width*1.618))
quadruple_fig.subplots(4,1)
for i in range(4): quadruple_fig.axes[i].set_visible(False)
```

```python
fig_width = 9
double_fig1 = plt.figure(figsize=(fig_width,fig_width*1.618/2))
double_fig2 = plt.figure(figsize=(fig_width,fig_width*1.618/2))
double_fig1.subplots(2,1)
double_fig2.subplots(2,1)
```

```python
# Plot the photon survival probability
ax_num = 0
A.configure_plot(xmin=xmin, xmax=xmax, )

A.compute_case(**{**pgg_arguments,
    'null':True,
    'color':'k',
    'axes_survival':quadruple_fig.axes[ax_num]
})
A.compute_case(**{**pgg_arguments,
    'null':True,
    'color':'k',
    'axes_survival':double_fig1.axes[ax_num]
})

A.configure_model(ALP_seed=demo_ALP_seed, params = demo_params) 
A.compute_case(**{**pgg_arguments,'axes_survival':quadruple_fig.axes[ax_num]})
A.compute_case(**{**pgg_arguments,'axes_survival':double_fig1.axes[ax_num]})
for i in range(4): quadruple_fig.axes[i].set_visible(False)
quadruple_fig.axes[ax_num].set_visible(True)
quadruple_fig
# A.fig_survival.savefig(thesis_figs_dir+'fig_residual_logs_pgg.jpg')
# A.fig_survival
```

```python
ax_num = 1
A.configure_plot(logy=False, xmin=xmin, xmax=xmax, ymin=None, ymax=None, legend_loc='upper right')

# Without ALPs, only expectations, not observations.
A.configure_model(params = [0,0], residuals=False, logcounts=True)
A.configure_obs(nbins=40)
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments,'axes':quadruple_fig.axes[ax_num]})
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments,'axes':double_fig1.axes[ax_num]})

# With ALPs, observations included.
A.configure_model(ALP_seed=demo_ALP_seed, params = demo_params)   
A.compute_case(**{**general_arguments,'axes':quadruple_fig.axes[ax_num]})
A.compute_case(**{**general_arguments,'axes':double_fig1.axes[ax_num]})
for i in range(4): quadruple_fig.axes[i].set_visible(False)
quadruple_fig.axes[ax_num].set_visible(True)
quadruple_fig
```

```python
ax_num = 2
A.configure_plot(logy=False, xmin=xmin, xmax=xmax, ymin=None, ymax=None, legend_loc='upper center')
A.configure_model(residuals=True, logcounts=False)

# Without ALPs, only expectations, not observations.
A.configure_model(params = [0,0])
A.configure_obs(nbins=40)
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments,'axes':quadruple_fig.axes[ax_num]})
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments,'axes':double_fig2.axes[ax_num-2]})

# With ALPs, observations included.
A.configure_model(ALP_seed=demo_ALP_seed, params = demo_params)   
A.compute_case(**{**general_arguments,'axes':quadruple_fig.axes[ax_num]})
A.compute_case(**{**general_arguments,'axes':double_fig2.axes[ax_num-2]})

for i in range(4): quadruple_fig.axes[i].set_visible(False)
quadruple_fig.axes[ax_num].set_visible(True)
quadruple_fig
# A.fig.savefig(thesis_figs_dir+'fig_residual_logs_residual.jpg')
```

```python
ax_num = 3
A.configure_plot(logy=False, xmin=xmin, xmax=xmax, ymin=-0.3, ymax=0.2, legend_loc='upper center')
A.configure_obs(nbins=40)
A.configure_model(residuals=True, logcounts=True)

# Without ALPs, only expectations, not observations.
A.configure_model(params = [0,0])
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments,'axes':quadruple_fig.axes[ax_num]})
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments,'axes':double_fig2.axes[ax_num-2]})


# With ALPs, observations included.
A.configure_model(ALP_seed=demo_ALP_seed, params = demo_params)   
A.compute_case(**{**general_arguments,'axes':quadruple_fig.axes[ax_num]})
A.compute_case(**{**general_arguments,'axes':double_fig2.axes[ax_num-2]})

for i in range(4): quadruple_fig.axes[i].set_visible(False)
quadruple_fig.axes[ax_num].set_visible(True)
quadruple_fig
# A.fig.savefig(thesis_figs_dir+'fig_residual_logs_residual_log.jpg')
```

```python
for fig in [double_fig1, double_fig2,quadruple_fig]:
    for i in range(len(fig.axes[:-1])): 
        fig.axes[i].set_xlabel(None)
        fig.axes[i].set_xticklabels([])
    for i in range(len(fig.axes)):
        fig.axes[i].set_visible(True)
        fig.axes[i].set_title(None)
        fig.axes[i].yaxis.set_label_coords(-0.12,0.5)
    fig.tight_layout()
double_fig1.savefig(thesis_figs_dir+"fig_residual_logs_double1.jpg")
double_fig2.savefig(thesis_figs_dir+"fig_residual_logs_double2.jpg")
quadruple_fig.savefig(thesis_figs_dir+"fig_residual_logs_quadruple.jpg")
```

```python
# Plot expected signals for different random B-field configurations

A.configure_plot(logy=False, xmin=xmin, xmax=xmax, ymin=-0.06, ymax=0.03, legend_loc='upper center')

A.configure_obs(nbins=90)
A.configure_model(residuals=True, logcounts=True, ALP_seed=None,params = demo_params)

# Without ALPs, only expectations, not observations.
A.compute_case(**{**general_arguments,**standard_hypothesis_arguments})

# With ALPs, observations included.
for color,linestyle in [('r','-'),('g','--'),('b',':')]:
    A.compute_case(**{**general_arguments,
        'errorbands':False,
        'plot_obs':False,
        'color':color,
        'linestyle':linestyle,
    })

A.fig.tight_layout()
A.fig.savefig(thesis_figs_dir+'fig_residual_logs_B_configs.jpg')
```
